```python
# تثبيت المكتبات اللازمة
!pip install google-genai pydantic gradio pillow -q

from google import genai
from google.genai import types
from google.colab import userdata
from pydantic import BaseModel
import gradio as gr
import os

# الاتصال بالنموذج باستخدام المفتاح المخزن في الـ Secrets
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)
MODEL = 'gemini-1.5-flash'
```

```python
class FoodItem(BaseModel):
    name: str                  # اسم عنصر الطعام (مثال: white rice)
    estimated_grams: int       # الوزن المقدر بناءً على الصورة
    calories_per_100g: int     # السعرات لكل 100 جرام
    confidence: str            # مستوى الثقة: "low", "medium", أو "high"

class MealAnalysis(BaseModel):
    is_food: bool              # هل هي وجبة طعام؟
    dish_name: str             # اسم الطبق (مثال: Mansaf)
    description: str           # وصف قصير للوجبة
    items: list[FoodItem]      # قائمة العناصر
    hidden_ingredients: list[str] # مكونات مخفية (مثل: زيت، زبدة)
```

```python
def analyze_meal(image, language="English"):
    prompt = (f"You are a nutrition assistant. Answer in {language}. "
              "First decide if this image shows food. "
              "If not, set is_food to false and leave items empty. "
              "If it is food, list EACH food item separately (rice, meat, sauce, bread...). "
              "Estimate the grams you can see, using the plate and cutlery for scale, "
              "and a typical calories_per_100g for that food as it is usually cooked. "
              "List likely hidden ingredients such as oil, butter or sugar. "
              "Set confidence to low when the food is covered, mixed or unclear.")

    response = client.models.generate_content(
        model=MODEL,
        contents=[image, prompt],
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=MealAnalysis
        )
    )
    return response.parsed.model_dump()
```

```python
def calculate(meal):
    rows = []
    for item in meal["items"]:
        kcal = round(item["estimated_grams"] * item["calories_per_100g"] / 100)
        rows.append([item["name"], item["estimated_grams"], kcal, item["confidence"]])

    total = sum(row[2] for row in rows)
    low, high = round(total * 0.75), round(total * 1.25) # نطاق بنسبة ±25%
    return rows, total, low, high
```

```python
def ask_meal(image, question, rows, total, language):
    table = "\n".join(f"- {name}: {grams}g, {kcal} kcal" for name, grams, kcal, _ in rows)
    prompt = (f"Estimated items in this meal:\n{table}\nTotal: {total} kcal.\n"
              f"Answer briefly in {language}, using only these numbers. "
              "If the answer can't be seen in the image, say so. "
              "Never call the food or the person good, bad or unhealthy.\n"
              f"Question: {question}")

    return client.models.generate_content(model=MODEL, contents=[image, prompt]).text
```

```python
import gradio as gr

def analyze(image, language):
    if image is None:
        return "Please upload a photo.", [], "", None
    try:
        meal = analyze_meal(image, language)
    except Exception as e:
        return f"Error: {e}", [], "", None

    if not meal["is_food"]:
        return "This doesn't look like food. Try another photo.", [], "", None

    rows, total, low, high = calculate(meal)
    summary = (f"### Total = {total} kcal (Range: {low}–{high})\n"
               "*Estimate only, not medical or diet advice.*")
    return f"{meal['dish_name']}: {meal['description']}", rows, summary, (rows, total)

def answer(image, question, saved, language):
    if image is None or saved is None:
        return "Please analyze a photo first."
    if not question or not question.strip():
        return "Please type a question."
    rows, total = saved
    try:
        return ask_meal(image, question, rows, total, language)
    except Exception as e:
        return f"Error: {e}"


custom_css = """
body {
    background-color: #0A0A0A !important;
    color: #FFFFFF !important;
}
.gradio-container {
    background-color: #0A0A0A !important;
    border-radius: 16px;
    padding: 20px;
}
/* تأثير الأزرار ثلاثية الأبعاد والتفاعل */
button.primary {
    background: linear-gradient(135deg, #007BA7, #005f7f) !important;
    border: none !important;
    color: white !important;
    font-weight: bold !important;
    box-shadow: 0px 6px 0px #003f53, 0px 10px 20px rgba(0, 123, 167, 0.4) !important;
    transition: all 0.2s ease-in-out !important;
    border-radius: 12px !important;
}
button.primary:hover {
    transform: translateY(-3px) !important;
    box-shadow: 0px 9px 0px #003f53, 0px 15px 25px rgba(255, 45, 149, 0.4) !important;
    background: linear-gradient(135deg, #0093c4, #007BA7) !important;
}
button.primary:active {
    transform: translateY(2px) !important;
    box-shadow: 0px 3px 0px #003f53 !important;
}

/* أزرار ثانوية بتأثير Neon Rose */
button:not(.primary) {
    background-color: #121212 !important;
    border: 2px solid #FF2D95 !important;
    color: #FF2D95 !important;
    font-weight: bold !important;
    box-shadow: 0px 4px 0px #80144a, 0px 6px 15px rgba(255, 45, 149, 0.2) !important;
    transition: all 0.2s ease-in-out !important;
    border-radius: 12px !important;
}
button:not(.primary):hover {
    background-color: #FF2D95 !important;
    color: #FFFFFF !important;
    transform: translateY(-2px) !important;
    box-shadow: 0px 6px 0px #80144a, 0px 10px 20px rgba(255, 45, 149, 0.5) !important;
}

/* تنسيق الصناديق والنصوص */
.gr-box, .gr-input, .gr-text-input, textarea {
    background-color: #141414 !important;
    border: 1px solid #262626 !important;
    color: #FFFFFF !important;
    border-radius: 10px !important;
}
h1 {
    color: #007BA7 !important;
    text-shadow: 0px 0px 15px rgba(0, 123, 167, 0.4);
    font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
}
"""

with gr.Blocks(title="Calorie Lens", css=custom_css) as demo:
    gr.Markdown("# 🥗 Calorie Lens - AI Nutrition Assistant")
    saved = gr.State()

    with gr.Row():
        with gr.Column():
            image_in = gr.Image(type="pil", label="Meal photo")
            language = gr.Radio(["English", "Arabic"], value="English", label="Language")
            analyze_btn = gr.Button("🔍 Analyze meal", variant="primary")
        with gr.Column():
            dish_out = gr.Textbox(label="Dish")
            table_out = gr.Dataframe(headers=["Food", "Grams", "kcal", "Confidence"], label="Items")
            total_out = gr.Markdown()

    question_in = gr.Textbox(label="Ask about this meal", placeholder="How many calories without the bread?")
    ask_btn = gr.Button("Ask Question")
    answer_out = gr.Textbox(label="Answer")

    analyze_btn.click(analyze, [image_in, language], [dish_out, table_out, total_out, saved])
    ask_btn.click(answer, [image_in, question_in, saved, language], answer_out)

demo.launch(share=True)
```
